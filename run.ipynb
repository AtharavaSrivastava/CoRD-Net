# CoRD-Net — FGBF Ablation Experiments (Google Colab T4 GPU)
Clones `https://github.com/sanjayrk2007/CoRD-Net-dev` (branch `dev`) and runs controlled training and evaluation for:
1. **E1-FGBF-Mild** (Baseline ConvNeXt + FGBF)
2. **E2-FGBF-Mild** (STN Auto-Localization + FGBF)
3. **E3-FGBF-Mild** (STN + Dual-Intensity + FGBF)

```python
%%bash
# Clone the dev branch of CoRD-Net-dev repository
if [ ! -d "CoRD-Net-dev" ]; then
  git clone -b dev https://github.com/sanjayrk2007/CoRD-Net-dev.git
fi
cd CoRD-Net-dev

python -m pip install --upgrade pip
pip install -r requirements.txt

python -c "import torch; print('PyTorch:', torch.__version__, '| CUDA:', torch.cuda.is_available(), '| GPU:', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'None')"
nvidia-smi

git log -1 --oneline
git status

mkdir -p logs checkpoints results
```

## 1. Experiment: E1-FGBF-Mild (Baseline ConvNeXt + FGBF)

```python
%%bash
cd CoRD-Net-dev || cd .
mkdir -p logs checkpoints results/e1_fgbf_mild

set -o pipefail
python train.py --exp e1_fgbf --data-root Dataset --pretrained --device cuda --amp --epochs 60 --patience 10 --batch-size 16 --loss-type ce --sampler none --augmentation mild --seed 42 --results-dir results/e1_fgbf_mild 2>&1 | tee logs/e1_fgbf_mild.txt
python evaluate.py --checkpoint checkpoints/e1_fgbf_best.pt --exp e1_fgbf --data-root Dataset --split test --device cuda --results-dir results/e1_fgbf_mild 2>&1 | tee logs/e1_fgbf_mild_eval.txt
python evaluate.py --checkpoint checkpoints/e1_fgbf_best.pt --exp e1_fgbf --data-root Dataset --split test --device cuda --results-dir results/e1_fgbf_mild --visualize 2>&1 | tee logs/e1_fgbf_mild_visualization.txt
```
